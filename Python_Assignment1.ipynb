### Strings

```python
#1.String Concatenation

string= "Hello "
name = input("Enter your Name: ")

result = string + name
print(result)

```

```python
string1 = ", welcome to Python programming"
result1 = result + string1
print(result1)
```

```python
#2.String Slicing and Indexing


print("First Character : ",result1[0])
print("Last Character : ",result1[-1])
print("First Five Characters : ",result1[0:6])
print("Last 11 Characters : ",result1[-11:])
print("Reverse:", result1[::-1])
print("Python:", result1[23:29])
```

```python
#3.String Methods

strM = ("Python beginner tutorial")
print(strM)
print("uppercase : ", strM.upper())
print("lowecase : ", strM.lower())
print("capitalize : ", strM.capitalize())
print("occurences : ", strM.count("t"))
print("Replacing:", strM.replace("Python", "Machine Learning"))
```

### Tuples

```python
tuple1 = (10,20,30)
tuple2 = (40,50,60)
print(tuple1)
print(tuple2)
```

```python
t_combine = tuple1 + tuple2
print(t_combine)
```

```python
print("repeated 3 time :",t_combine*3)
print("third element :",t_combine[2])
print("first 3 elements :",t_combine[0:3])
print("last 3 elements :",t_combine[-3:])
```
